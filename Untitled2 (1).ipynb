```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from scipy.cluster.hierarchy import dendrogram, linkage
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, callbacks
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('Global_Pollution_Analysis - Global_Pollution_Analysis.csv')
print(df.shape)
df.head()
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
le = LabelEncoder()
df['Country_Code'] = le.fit_transform(df['Country'])
```

```python
df['Total_Pollution_Index'] = (df['Air_Pollution_Index'] + 
                                df['Water_Pollution_Index'] + 
                                df['Soil_Pollution_Index']) / 3
```

```python
features_for_clustering = ['Air_Pollution_Index', 'Water_Pollution_Index', 
                           'Soil_Pollution_Index', 'Energy_Recovered (in GWh)',
                           'CO2_Emissions (in MT)', 'Renewable_Energy (%)',
                           'Energy_Consumption_Per_Capita (in MWh)',
                           'Total_Pollution_Index']

scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[features_for_clustering])
scaled_df = pd.DataFrame(scaled_features, columns=features_for_clustering)
```

```python
inertia = []
K_range = range(1, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K_range, inertia, 'bo-')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method for K‑Means')
plt.show()
```

```python
k = 3
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df['KMeans_Cluster'] = kmeans.fit_predict(scaled_features)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_features = pca.fit_transform(scaled_features)

plt.figure(figsize=(8,6))
sns.scatterplot(x=pca_features[:,0], y=pca_features[:,1], 
                hue=df['KMeans_Cluster'], palette='Set1')
plt.title('K‑Means Clusters (PCA projection)')
plt.show()
```

```python
cluster_means = df.groupby('KMeans_Cluster')[features_for_clustering].mean()
print(cluster_means)
```

```python
linked = linkage(scaled_features, method='ward')
plt.figure(figsize=(12,6))
dendrogram(linked, truncate_mode='level', p=5, leaf_rotation=90.)
plt.title('Hierarchical Clustering Dendrogram')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=k, metric='euclidean', linkage='ward')
df['Hier_Cluster'] = hc.fit_predict(scaled_features)
```

```python
# Contingency table
pd.crosstab(df['KMeans_Cluster'], df['Hier_Cluster'])
```

```python
X = df[['Air_Pollution_Index', 'CO2_Emissions (in MT)', 'Industrial_Waste (in tons)',
        'Renewable_Energy (%)', 'Energy_Consumption_Per_Capita (in MWh)',
        'Population (in millions)', 'GDP_Per_Capita (in USD)']]
y = df['Energy_Recovered (in GWh)']

# Scale features
from sklearn.model_selection import train_test_split
scaler_X = StandardScaler()
X_scaled = scaler_X.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)   # output layer
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])
history = model.fit(X_train, y_train, epochs=100, batch_size=16, 
                    validation_split=0.2, verbose=1)
```

```python
y_pred = model.predict(X_test)
print(f"R²: {r2_score(y_test, y_pred):.3f}")
print(f"MSE: {mean_squared_error(y_test, y_pred):.3f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred):.3f}")
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='val')
plt.title('Loss')
plt.legend()
plt.subplot(1,2,2)
plt.plot(history.history['mae'], label='train')
plt.plot(history.history['val_mae'], label='val')
plt.title('MAE')
plt.legend()
plt.show()
```

```python
model2 = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dropout(0.3),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])

model2.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001),
               loss='mse', metrics=['mae'])

early_stop = callbacks.EarlyStopping(patience=20, restore_best_weights=True)

history2 = model2.fit(X_train, y_train, epochs=200, batch_size=16,
                      validation_split=0.2, callbacks=[early_stop], verbose=1)

y_pred2 = model2.predict(X_test)
print(f"Improved R²: {r2_score(y_test, y_pred2):.3f}")
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
print(f"Linear Regression R²: {r2_score(y_test, y_pred_lr):.3f}")
```

# Model Comparison

Clustering (K‑Means vs Hierarchical): Both gave similar cluster assignments; K‑Means is more scalable, Hierarchical provides a dendrogram that reveals nested relationships.

Neural Network vs Linear Regression: The neural network (after tuning) likely achieves better R², especially if there are non‑linear interactions between pollution and energy recovery. However, linear regression may be simpler and interpretable.

# Actionable Insights

From the clusters, we can identify groups of countries:

Cluster 0: High pollution, low energy recovery → need urgent pollution control and investment in energy recovery infrastructure.

Cluster 1: Moderate pollution, moderate recovery → balanced approach.

Cluster 2: Low pollution, high recovery → could serve as models for others.
